### Python Datatypes in Bioinformatics \& Mathematical Modeling

| Datatype | When to Use (Domain-Specific) | Example (Math/Bio Context) |
| :-- | :-- | :-- |
| **int** (integer) | Use for whole counts in genomics, population genetics, or item indexing. Good for representing discrete biological units like number of exons, individuals, or SNPs. | Counting mutations in a gene: `mutations = 12` <br> Using in loop: `for snp_id in range(mut_​ations): ...` |
| **float** (decimal) | Use for continuous measurements: probabilities, normalized expression levels, rate constants, p-values, or concentrations. | Modeling allele frequency after selection: `freq_A = 0.73` <br> Growth rate constant in microbial population: `r = 0.015` |
| **str** (string) | Use for categorical/sequence data: DNA, RNA, protein sequences, gene IDs, sample names. Strings are central for storing biological sequences. | DNA subsequence extraction: `dna = "ATGGCCATTGTAATGGGCC"` <br> Extract codon \#2: `codon2 = dna[3:6]  # "GCC"` |
| **bool** (boolean) | Use for filtering data, logical conditions, and building biological/mathematical rules. | Check if a mutation changes a stop codon: <br> `is_stop = codon2 in ["TAA","TAG","TGA"]  # False` |
| **list** | Use for ordered collections of values: replicate data, gene expression arrays, simulation results. Flexible and modifiable structure. | Store expression levels of BRCA1 across 5 patients: <br> `brca1_expr = [12.1, 15.4, 9.8, 22.0, 13.2]` <br> Average expression: `sum(brca1_expr)/len(brca1_expr)` |
| **tuple** | Use for representing fixed biological coordinates or immutable groupings (cannot be changed). Common in genome annotation or graph edges in networks. | Store gene location on chromosome: <br> `gene_loc = ("chr17", 43044295, 43125483)` <br> Gene span: `gene_loc[2]-gene_loc[1]` |
| **dict** (dictionary) | Use for mappings: gene ID → expression value, protein → function, SNP → effect. Fast lookups are ideal with large datasets. | Mapping multiple genes to TPM values: <br> `expr_dict = {"BRCA1": 12.1, "TP53": 22.5, "EGFR": 9.2}` <br> Access BRCA1: `expr_dict["BRCA1"]` |
| **set** | Use for representing unique biological categories: nucleotides in a sequence, unique species, or shared SNP sets. Great for intersections \& unions. | Find nucleotides in a sequence: <br> `seq = "ATGCGTACGTA"` <br> `unique_bases = set(seq)  # {'A','T','G','C'}` <br> Compare with another sequence: `common = unique_bases & set("TAGCTA")` |


- **Biologists** should think: *counts → int, sequences → str, experiments → list, lookups → dict, uniqueness → set*.
- **Mathematicians** should think: *discrete vs continuous models (int vs float), states/conditions (bool), fixed vs dynamic structures (tuple vs list), mappings (dict), and sets for intersections/unions or vector spaces*.

````python
# this looks like a dictionary representation but in a string

code = "'ATA':'I', 'ATC':'I', 'ATT':'I', 'ATG':'M',\
    'ACA':'T', 'ACC':'T', 'ACG':'T', 'ACT':'T',\
    'AAC':'N', 'AAT':'N', 'AAA':'K', 'AAG':'K',\
    'AGC':'S', 'AGT':'S', 'AGA':'R', 'AGG':'R',\
    'CTA':'L', 'CTC':'L', 'CTG':'L', 'CTT':'L',\
    'CCA':'P', 'CCC':'P', 'CCG':'P', 'CCT':'P',\
    'CAC':'H', 'CAT':'H', 'CAA':'Q', 'CAG':'Q',\
    'CGA':'R', 'CGC':'R', 'CGG':'R', 'CGT':'R',\
    'GTA':'V', 'GTC':'V', 'GTG':'V', 'GTT':'V',\
    'GCA':'A', 'GCC':'A', 'GCG':'A', 'GCT':'A',\
    'GAC':'D', 'GAT':'D', 'GAA':'E', 'GAG':'E',\
    'GGA':'G', 'GGC':'G', 'GGG':'G', 'GGT':'G',\
    'TCA':'S', 'TCC':'S', 'TCG':'S', 'TCT':'S',\
    'TTC':'F', 'TTT':'F', 'TTA':'L', 'TTG':'L',\
    'TAC':'Y', 'TAT':'Y', 'TAA':'_', 'TAG':'_',\
    'TGC':'C', 'TGT':'C', 'TGA':'_', 'TGG':'W'"
````

````python
# we can use eval to set it up as a dictionary

codons_dict = eval("{" + code + "}")
codons_dict
````

````python
# with a for loop
# set up the sequence
DNA_sequence = "GCCCGGCGTAAAAAT"

# set up the list of amino acids to an empty list
aa_list = []
# set the length of the DNA sequence in a variable
n = len(DNA_sequence)

# we need indices to get groups of 3 nucleotides
# so we create a range
for i in range(0,n,3):
    # get groups of 3 nuclotides using a start index i
    codon = DNA_sequence[i:i+3]
    # print(codon)

    # get the amino acid (value) for the codon (key)
    # from the dictionary
    # if the codon does not have an amino acid retrurn ""
    aa = codons_dict.get(codon,"")
    # print(aa)
    # add amino acid to the list
    aa_list.append(aa)

translated_peptide = "".join(aa_list)
translated_peptide
print("the peptide is:", translated_peptide)
````

##### Gene regulatory network

"Formally speaking, a gene regulatory network or genetic regulatory network (GRN) is a collection of DNA segments in a cell which interact with each other (indirectly through their RNA and protein expression products) and with other substances in the cell, thereby governing the rates at which genes in the network are transcribed into mRNA. In general, each mRNA molecule goes on to make a specific protein (or set of proteins)."  
https://link.springer.com/referenceworkentry/10.1007%2F978-1-4419-9863-7_364

<img src = "https://media.springernature.com/original/springer-static/image/prt%3A978-1-4419-9863-7%2F7/MediaObjects/978-1-4419-9863-7_7_Part_Fig1-364_HTML.gif" width = 200/>




We have a gene regulatory network represented as a dictionary where the key is a gene from the network above and the value is a tuple of the genes the key gene directly regulates (through orange links only).    

- Write code that uses the dictionary and computes and returns the number of genes that a given gene's expression is regualted by. 
    - In other words, the code would compute the number of incoming orange edges that the gene has in the above network.<BR><BR>

- Test the code for at least three cases.
- Put the code into a function and test the function - make the result a dictionary




````python
network = {'Gene1': ('Gene2',),
 'Gene2': ('Gene6',),
 'Gene3': ('Gene1', 'Gene5'),
 'Gene4': ('Gene2',),
 'Gene5': ('Gene1',),
 'Gene6': ('Gene1',)}
````

````python
network
````

````python
# we set up our output to 0, we assume we find no incoming edges
reg_no = 0

# set up the test the code
gene = "Gene1"
gene
````

````python
# we set up our output to 0, we assume we find no incoming edges
# we go through the dictionary items and display the key and value for each item

reg_no = 0
# the for loop through items will give us tuples of 2
# we unpack the key and value in the respective variables
for key, value in network.items():
    # here we will compute the number
    print("Displaying the key and value:")
    print(key)
    print(value)
    print()
print(reg_no)

````

````python
# we set up our output to 0, we assume we find no incoming edges
# we go through the dictionary items and display the key and value for each item
# we check if our gene is among the values for each item because that means it has an incoming edge from the key

reg_no = 0
# the for loop through items will give us tuples of 2
# we unpack the key and value in the respective variables
for key, value in network.items():
    # here we are computing the number
    # we check if our gene is among the gene in the value and 
    # if it is there means there is an incoming edge from the key so we increase the number 
    if gene in value:
        reg_no = reg_no + 1
print(reg_no)

````

````python
## testing the code with more values

test_genes = ("Gene1", "Gene2", "Gene4", "Gene6")
for gene in test_genes:
    reg_no = 0
    for key, value in network.items():
        # here we are computing the number
        # we check if our gene is among the gene in the value and 
        # if it is there means there is an incoming edge from the key so we increase the number 
        if gene in value:
            reg_no = reg_no + 1
print("The number of incoming edges for '" + gene + "' is:", reg_no)
````

````python
# Put the code in a function
# The result needs to be returned using return and 
# the function needs to be called using variables set up outside of the function 
# we set up our output to 0, we assume we find no incoming edges
# we go through the dictionary items and display the key and value for each item
# we check if our gene is among the values for each item because that means it has an incoming edge from the key

def compute_reg_no(gene, biological_network):
    reg_no = 0
    # the for loop through items will give us tuples of 2
    # we unpack the key and value in the respective variables
    for key, value in biological_network.items():
        # here we are computing the number
        # we check if our gene is among the gene in the value and 
        # if it is there means there is an incoming edge from the key so we increase the number 
        if gene in value:
            reg_no = reg_no + 1
    return reg_no

# test the function
test_gene = "Gene1"
res = compute_reg_no(gene = test_gene, biological_network = network)
res
````

````python
## Saving the results in a dictionary

edge_count = {}

test_genes = ("Gene1", "Gene2", "Gene4", "Gene6")
for test_gene in test_genes:
    incoming_edges = compute_reg_no(gene = test_gene, biological_network = network)
    edge_count[test_gene] = incoming_edges

edge_count
````

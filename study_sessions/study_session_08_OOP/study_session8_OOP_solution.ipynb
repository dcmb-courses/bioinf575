#### Implement the following classes: Gene, GOTerm, GENEGOTerm

```python
# implement the Gene class
# implement the __init__ method 
# in __init__ we create custom attributes:  gid, symbol and description
# implement __str__ and __repr__ methods to be able to display the object 
# implement the __eq__ method to be able to chech equality between Gnee objects

class Gene:
    def __init__(self,  gid, sym, desc): 
        # checks can be done to make sure the data is not missing and 
        # each parameter should ahve the correct type
        self.gid = gid
        self.symbol = sym
        self.description = desc
        
    def __str__(self):
        return f"Gene({self.gid},'{self.symbol}','{self.description}')"
    
    def __repr__(self):
        return f"Gene({self.gid},'{self.symbol}','{self.description}')"
    
    def __eq__(self, g):
        return self.gid == g.gid and self.symbol == g.symbol and self.description == g.description
    
    
```

#### Create a list of genes, a list of GO terms, and a list of gene to GO term elements from the file `genes_info1.txt`

```python
# It should have the following method: is_indel

# we add a new function definition
# this method will be called from a Variant object so it needs the self parameter

class Variant:
    def __init__(self, chr, p, id, ref, alt, q):
        self.CHROM = chr
        self.POS = p
        self.ID = id
        self.REF = ref
        self.ALT = alt
        self.QUAL = q

    def __str__(self):
        # using the + operator to create the string we want to display on str() or print()
        # we also use str for any attribute that is not a string POS & QUAL
        vstr = self.CHROM + " " + str(self.POS) + " " + self.ID + " " + self.REF + " " + self.ALT + " " + str(self.QUAL) 
        return vstr

    def __repr__(self):
        # we can use the + operator to create the string 
        # but an f-string is easier
        # this string when using eval on it it should create new variant
        # it has to have the exact syntax of calling the Variant()/__init__ function withthe correct parameters
        # including the quotes for string parameters like CHROM or REF
        vstr = f"Variant('{self.CHROM}', {self.POS}, '{self.ID}', '{self.REF}', '{self.ALT}', {self.QUAL})" 
        return vstr

    def is_indel(self):
        """
        Arguments: 
            self: a variant object
        Returns: 
            True/False - a response to the question is the variant an indel
            An indel is an insertion or deletion - which means 
            the reference and alternative for this variant 
            do not have the same length
        """
        return len(self.REF) != len(self.ALT)
```

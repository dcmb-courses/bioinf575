### The Big Idea 
> The idea behind objects is to **bundle** coherent <u>methods</u> (things the object can _do_) and <u>attributes</u> (things the object _has_) that logically go together into a well-defined _interface_.

They are a data abstraction that has 2 main jobs:
1. Captures internal *representation* of the data it is abstracting
2. Creates an *interface* for the abstracted data

#### Let's think about a Pokemon and basic information we want to store about it.
We could make a dictionary:

```python

{"Name": "Pikachu", "Level" = 5}

```


While this structure works perfectly fine, if we want to add new elements or add a function specific to the Pokemon, it will be hard to keep track of it all.
It makes sense that all of these things could be wrapped up into a single object (_mainly because it is hard to manage and add new funtionality to it_).

#### Let's make a `Pokemon` object 

- <font color = "red">Use the `__init__` method to define class variables/ attributes</font>
- Write functions in the class for extra functionality
- Use `self` to refer to the object you are creating



````python
class Pokemon:
    def __init__(self, name = "Pikachu", level = 1):
        self.name = name
        self.level = level

    def __str__(self):
        return self.name

    def get_level(self):
        return self.level
````

---
We need to take a second to talk about 3 things real quick:
1. Functions within `class`es (like `self.get_level`) are called ***methods*** or procedural attributes
2. `self.name` and `self.level` are called ***attributes*** since they only contain data
3. What in the world is `self`?

**PS**: `self` is a parameter that allows an object to look back at its self. Specifically the current *instance* of itself. However, outside of writing the class, you will never actually have to pass the word `self` into the methods.

#### Let's add more functionality to the `Pokemon` class
- Implement specific dunder methods for added functionality 
    - the dunder method name is representative for the function/ operation it is used for
- Use @property to set up read-only attributes
- Add documentation using docstrings

````python
import random

# Pokemon object
class Pokemon:
    """
    Contains the information about a Pokemon such as its name, type,
    level, stats, and moves. 
    
    Attributes:
    name (str):  Pokemon name given by user
    __level (int): Pokemon level
    __base_stats (dict): Pokemon base stats (e.g., HP, Attack, Defense)
    """
    def __init__(self, pname = "Pikachu", plevel = 1):
        self.name = pname
        self.__level = plevel
        self.__base_stats = {"base_HP": random.randint(33,75), 
                             "base_Attack": random.randint(33,75), 
                             "base_Defense": random.randint(33,75),
                             "base_spAttack": random.randint(33,75),
                             "base_spDefense": random.randint(33,75),
                             "base_Speed": random.randint(33,75)}
        
    def __str__(self):
        return f"Pokemon('{self.name}',{self.__level})"
    
    def __repr__(self):
        return f"Pokemon('{self.name}',{self.__level})"
    
    def __eq__(self, other):
        return self.name == other.name and self.__level == other.__level

        
    @property # getter
    def base_stats(self):
        """
        Get the stats for the Pokemon
        """
        return self.__base_stats
    
    @property # getter
    def level(self):
        """
        Get the level of the Pokemon
        """
        return self.__level

    # @level.setter # property setter - same name as the property
    # def update_level(self, plevel):
    #    self.__level = plevel
             
    def level_up(self):
        """
        Level up your Pokemon!
        """
        if self.__level < 100:
            self.__level += 1
        
        
````

````python
# Use Pokemon - explore
# create/init Pokemon objects, we need the __init__ method
# The constructor without arguments works if we have default values
# The cell will use the __repr__ method to display the representation of the object

p = Pokemon()
p
````

#### Post-class notes

 * While I added new functionality to the `Pokemon` class method `level_up()` and reran that cell, I didn't run the subsequent code `p = Pokemon()` to re-initiate the new `Pokemon` object assigned to variable `p`. In other words, `p` was not updated according to my new code! 
 If you rerun the code in the order of this notebook, levelling up `p` will max out at 100 instead of going up to level 205, because you will not be using the old `p` (before the max level of 100 was implemented in the `level_up()` method). 

 * _Properties_ don't necessarily have to be _attributes_, though properties can use attributes to compute something internally before outputting the intended property.
    * Example: let's say I wanted my `Pokemon` class to have a property called current hit points: `current_HP` $= 2 \times \text{base\_stat} + 10 + \text{level}$ (for simplicity; the game's formula is more complicated). Storing this value as an attribute would be cumbersome, as you would constantly have to update it during level up. Instead, you can create a property to compute `current_HP` each time you need to use it:

        ```python
        @property
        def current_HP(self):
            return 2*self.base_stats{"base_Attack"} + 10 + self.__level
        ```
        You can then call `p.current_HP` to output the Pokemon's current hit points, which will depend on the `Pokemon` object's base stats and level.
